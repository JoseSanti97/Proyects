```python
ruta_1 = '/content/clientes.csv'
ruta_2 = '/content/ml_part1.csv'
ruta_3 = '/content/ml_part2.csv'
ruta_4 = '/content/transacciones.csv'

df_1 = pd.read_csv(ruta_1)
df_2 = pd.read_csv(ruta_2)
df_3 = pd.read_csv(ruta_3)
df_4 = pd.read_csv(ruta_4)

print(df_4.head())
```

```python
df_1 = df_1.drop(columns=['nombre', 'apellido', 'curp', 'ciudad', 'correo'])
print(df_1.head())
```

```python
df_ml= df_2.rename(columns={"cliente_id": "id_cliente"})
print(df_ml.head())
```

```python
df_ml= pd.merge(df_ml, df_1, on='id_cliente', how= 'inner')
df_ml = pd.merge(df_ml, df_4, on='id_cliente', how = 'inner')
print(df_ml.head())
```
